```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, f1_score, recall_score, accuracy_score,confusion_matrix
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
import numpy as np
```

```python
# Data Selection
data=pd.read_csv("../Data/processed_data.csv")
data.info()
```

```python
# Split data
X=data.drop("Loan_Approved",axis=1)
Y=data["Loan_Approved"]
```

```python
# Convert data into Train And Test Data
X_train,X_test,Y_train,Y_test=train_test_split(
    X,
    Y,
    test_size=.2,
    random_state=42
)
```

```python
# feature scaling
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
# Model Selction,Training,Prediction
knnModel=KNeighborsClassifier(n_neighbors=5)
knnModel.fit(X_train_scaled,Y_train)
Y_pred=knnModel.predict(X_test_scaled)
```

```python
# Model Evoluation
print("Precision Score:",precision_score(Y_test,Y_pred)*100,"%")
print("Recall Score:",recall_score(Y_test,Y_pred)*100,"%")
print("F1 Score:",f1_score(Y_test,Y_pred)*100,"%")
print("Precision Score:",accuracy_score(Y_test,Y_pred)*100,"%")
print("ConfusionMetrix Score:",confusion_matrix(Y_test,Y_pred))
```

```python
data["DTI_Ratio_Squre"]=data["DTI_Ratio"]**2
data["Credit_Score_Square"]=data["Credit_Score"]**2
data["Applicant_Income_Log"]=np.log1p(data["Applicant_Income"])

X=data.drop(columns=["Loan_Approved","DTI_Ratio","Credit_Score","Applicant_Income"])
Y=data["Loan_Approved"]

X_train,X_test,Y_train,Y_test=train_test_split(
    X,
    Y,
    test_size=.2,
    random_state=42
)
# feature scaling
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)

# Model Selction,Training,Prediction
knnModel=KNeighborsClassifier(n_neighbors=5)
knnModel.fit(X_train_scaled,Y_train)
Y_pred=knnModel.predict(X_test_scaled)

# Model Evoluation
print("Precision Score:",precision_score(Y_test,Y_pred)*100,"%")
print("Recall Score:",recall_score(Y_test,Y_pred)*100,"%")
print("F1 Score:",f1_score(Y_test,Y_pred)*100,"%")
print("Precision Score:",accuracy_score(Y_test,Y_pred)*100,"%")
print("ConfusionMetrix Score:",confusion_matrix(Y_test,Y_pred))
```
