```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import precision_score, f1_score, recall_score, accuracy_score,confusion_matrix
from sklearn.linear_model import LogisticRegression
from sklearn.preprocessing import StandardScaler
import numpy as np
```

```python
detaSet=pd.read_csv("../Data/processed_data.csv")
detaSet.info()
```

```python
X=detaSet.drop("Loan_Approved",axis=1)
Y=detaSet["Loan_Approved"]
```

```python
X_train,X_test,Y_train,Y_test=train_test_split(
    X,
    Y,
    train_size=.2,
    random_state=42
)
```

```python
# feature scaling
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)
```

```python
# model selction and training
logModel=LogisticRegression()
logModel.fit(X_train_scaled,Y_train)
Y_pred=logModel.predict(X_test_scaled)
```

```python
#model evaluation: 1st priority-precision score, 2nd priority-Recall score
print("Precision Score:",precision_score(Y_test,Y_pred)*100,"%")
print("Recall Score:",recall_score(Y_test,Y_pred)*100,"%")
print("F1 Score:",f1_score(Y_test,Y_pred)*100,"%")
print("Precision Score:",accuracy_score(Y_test,Y_pred)*100,"%")
print("ConfusionMetrix Score:",confusion_matrix(Y_test,Y_pred))
```

```python
#feature Engineering
detaSet["DTI_Ratio_Squre"]=detaSet["DTI_Ratio"]**2
detaSet["Credit_Score_Square"]=detaSet["Credit_Score"]**2
detaSet["Applicant_Income_Log"]=np.log1p(detaSet["Applicant_Income"])

X=detaSet.drop(columns=["Loan_Approved","DTI_Ratio","Credit_Score","Applicant_Income"])
Y=detaSet["Loan_Approved"]

X_train,X_test,Y_train,Y_test=train_test_split(
    X,
    Y,
    test_size=.2,
    random_state=42
)
# feature scaling
scaler=StandardScaler()
X_train_scaled=scaler.fit_transform(X_train)
X_test_scaled=scaler.transform(X_test)

# Model Selction,Training,Prediction
logModel=LogisticRegression()
logModel.fit(X_train_scaled,Y_train)
Y_pred=logModel.predict(X_test_scaled)

# Model Evoluation
print("Precision Score:",precision_score(Y_test,Y_pred)*100,"%")
print("Recall Score:",recall_score(Y_test,Y_pred)*100,"%")
print("F1 Score:",f1_score(Y_test,Y_pred)*100,"%")
print("Precision Score:",accuracy_score(Y_test,Y_pred)*100,"%")
print("ConfusionMetrix Score:",confusion_matrix(Y_test,Y_pred))
```
